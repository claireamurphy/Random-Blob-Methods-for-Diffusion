# Imports

```python
import numpy as np
from matplotlib import pyplot as plt

from src.initial_data import initial_data
from src.integrators import solve_ODE
from src.error import wass_2_approx
```

# Configuration

```python
m = 100
h = .02

moll_eps = 4 * h**(.99)

t_max = 1.05
t_step = .01

alg = 2
num_batches = 2
ratio = 2
```

# Function Definitions


```python
def V(x, y):
    return (x**2 + y**2) / 2

# Gradient of velocity field
def grad_V(x, y):
    return np.concatenate([x, y])
    
# Distribution of our initial data
def rho_0(x, y):
    R = np.e / 2
    return np.where(abs(x) <= R and abs(y) <= R, 1/(2 * R)**2, 0)
  
# 2nd derivative of Moreau-Yosida envelope
def f_double_prime_epsilon(x):
    return m * x**(m - 2)

def func(t, x, y):
    if t <= 1:
        return np.exp(2 * t) * rho_0(np.exp(t) * x, np.exp(t) * y)
    else:
        return np.exp(2) * rho_0(np.exp(1) * x, np.exp(1) * y)
```

# Main


```python
x0, y0, m_vals = initial_data(rho_0, np.e / 2 + h, h)
print("N = " + str(m_vals.size) + ".")

particle_sol, runtime = solve_ODE(x0, y0, m_vals, alg, num_batches, ratio, moll_eps, f_double_prime_epsilon, grad_V, t_step, t_max, bounds = 3) 

def func(x, y):
    if t_max <= 1:
        return np.exp(2 * t_max) * rho_0(np.exp(t_max) * x, np.exp(t_max) * y)
    else:
        return np.exp(2) * rho_0(np.exp(1) * x, np.exp(1) * y)
print(wass_2_approx(particle_sol[-1], m_vals, func, 2, .02))
```
