# Imports

```python
import numpy as np
from matplotlib import pyplot as plt

from src.initial_data import initial_data
from src.integrators import solve_ODE
from src.error import wass_2_approx
```

# Configuration

```python
m = 5
h = .01

moll_eps = 4*h**(.99)

t_step = 0.00625
t_max = .1

alg = 2
num_batches = 2
ratio = 2
```

# Function Definitions


```python
# Gradient of velocity field
def grad_V(x, y):
    return 0
    
# Normalizing constant in our Barenblatt function
def normalizing_const():
  if m == 5:
     return .03643

# Distribution of our initial data
def rho_0(x, y):
    return psi(0, x, y)
  
def tau():
   if m == 1:
      return 1 / (4 * np.pi)
   if m != 1:
      beta = 1 / (2 * m - 1)
      return normalizing_const()**( 1 / (2 * beta * (m - 1)) )

# Gaussian or Barenblatt
def psi(t, x, y):
  if m == 1:
      denom = 4 * np.pi * (tau() + t)
      arg = - (x**2 + y**2) / (4 * (tau() + t))
      return np.exp(arg) / denom

  else:
    beta = 1 / (2 * m)
    kappa = beta / 2 * (m - 1) / m
    K = normalizing_const() 
    temp = K - kappa * (t + tau())**(-2 * beta) * (x**2 + y**2)
    return (tau() + t)**(-2 * beta) * np.maximum(0, temp)**(1/(m - 1))
  
# 2nd derivative of Moreau-Yosida envelope
def f_double_prime_epsilon(x):
    return m * x**(m - 2)
```

# Main


```python
# Initialize data
x0, y0, m_vals = initial_data(rho_0, .7, h)
print("N = " + str(m_vals.size) + ".")

particle_sol, runtime = solve_ODE(x0, y0, m_vals, alg, num_batches, ratio, moll_eps, f_double_prime_epsilon, grad_V, t_step, t_max) 

def func(x, y):
    return psi(t_max, x, y)
print(wass_2_approx(particle_sol[-1], m_vals, func, .8, .015))
```
